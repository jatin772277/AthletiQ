```python
import numpy as np
import cv2

def load_yolo_labels(label_path):
    boxes = []

    if not label_path.exists():
        return np.empty((0, 4), dtype=np.float32)

    with open(label_path, "r") as f:
        for line in f:
            parts = line.strip().split()

            if len(parts) != 5:
                continue

            class_id = int(parts[0])

            if class_id != 0:
                continue

            xc, yc, w, h = map(float, parts[1:])

            boxes.append([xc, yc, w, h])

    if not boxes:
        return np.empty((0, 4), dtype=np.float32)

    return np.array(boxes, dtype=np.float32)


def yolo_to_xyxy(boxes, image_width, image_height):
    if len(boxes) == 0:
        return np.empty((0, 4), dtype=np.float32)

    result = []

    for xc, yc, w, h in boxes:
        x1 = (xc - w / 2) * image_width
        y1 = (yc - h / 2) * image_height
        x2 = (xc + w / 2) * image_width
        y2 = (yc + h / 2) * image_height

        result.append([x1, y1, x2, y2])

    return np.array(result, dtype=np.float32)


def calculate_iou(box1, box2):
    x1 = max(box1[0], box2[0])
    y1 = max(box1[1], box2[1])
    x2 = min(box1[2], box2[2])
    y2 = min(box1[3], box2[3])

    iw = max(0, x2 - x1)
    ih = max(0, y2 - y1)

    intersection = iw * ih

    area1 = max(0, box1[2] - box1[0]) * max(
        0, box1[3] - box1[1]
    )

    area2 = max(0, box2[2] - box2[0]) * max(
        0, box2[3] - box2[1]
    )

    union = area1 + area2 - intersection

    if union <= 0:
        return 0.0

    return intersection / union


def greedy_match(gt_boxes, pred_boxes, iou_threshold=0.5):

    if len(gt_boxes) == 0:
        return 0, len(pred_boxes), 0, []

    if len(pred_boxes) == 0:
        return 0, 0, len(gt_boxes), []

    pairs = []

    for gt_idx, gt_box in enumerate(gt_boxes):
        for pred_idx, pred_box in enumerate(pred_boxes):

            iou = calculate_iou(gt_box, pred_box)

            if iou >= iou_threshold:
                pairs.append(
                    (iou, gt_idx, pred_idx)
                )

    pairs.sort(reverse=True)

    matched_gt = set()
    matched_pred = set()
    matched_ious = []

    for iou, gt_idx, pred_idx in pairs:

        if gt_idx in matched_gt:
            continue

        if pred_idx in matched_pred:
            continue

        matched_gt.add(gt_idx)
        matched_pred.add(pred_idx)

        matched_ious.append(iou)

    tp = len(matched_gt)
    fp = len(pred_boxes) - tp
    fn = len(gt_boxes) - tp

    return tp, fp, fn, matched_ious


# ----------------------------------
# Test first validation image
# ----------------------------------

test_image = val_images[0]

print("Testing image:")
print(test_image)

# Run YOLO
results = model.predict(
    source=str(test_image),
    conf=CONF_THRESHOLD,
    imgsz=IMAGE_SIZE,
    device=0,
    verbose=False
)

result = results[0]

# Read image
image = cv2.imread(str(test_image))
height, width = image.shape[:2]

# Predictions
pred_boxes = result.boxes.xyxy.cpu().numpy()

# Ground truth
label_path = (
    VAL_LABELS
    / test_image.parent.name
    / f"{test_image.stem}.txt"
)

gt_yolo = load_yolo_labels(label_path)

gt_boxes = yolo_to_xyxy(
    gt_yolo,
    width,
    height
)

# Match
tp, fp, fn, matched_ious = greedy_match(
    gt_boxes,
    pred_boxes,
    IOU_THRESHOLD
)

print("\n========== SINGLE IMAGE TEST ==========")
print("Image size :", width, "x", height)
print("GT players:", len(gt_boxes))
print("Predictions:", len(pred_boxes))
print("TP:", tp)
print("FP:", fp)
print("FN:", fn)

if matched_ious:
    print(
        "Mean IoU:",
        round(float(np.mean(matched_ious)), 4)
    )

if len(result.boxes.conf) > 0:
    print(
        "Mean confidence:",
        round(
            float(result.boxes.conf.mean().item()),
            4
        )
    )

# GPU memory
import torch

if torch.cuda.is_available():
    print(
        "GPU memory allocated:",
        round(
            torch.cuda.memory_allocated() / (1024**3),
            2
        ),
        "GB"
    )

    print(
        "GPU memory reserved:",
        round(
            torch.cuda.memory_reserved() / (1024**3),
            2
        ),
        "GB"
    )
```

```python
import pandas as pd
import numpy as np
import shutil
import time
import torch
from pathlib import Path
from ultralytics import YOLO

# ============================================================
# HARD MINING
# ============================================================

print("========== STARTING HARD MINING ==========")

# ------------------------------------------------------------
# Load existing checkpoint if available
# ------------------------------------------------------------

if CHECKPOINT_DRIVE.exists():

    print("Found Drive checkpoint.")
    print("Loading previous progress...")

    shutil.copy2(
        CHECKPOINT_DRIVE,
        CHECKPOINT_LOCAL
    )

elif CHECKPOINT_LOCAL.exists():

    print("Found local checkpoint.")
    print("Resuming from local progress...")

else:

    print("No previous checkpoint found.")
    print("Starting from image 1.")

# ------------------------------------------------------------
# Load completed images
# ------------------------------------------------------------

if CHECKPOINT_LOCAL.exists():

    checkpoint_df = pd.read_csv(
        CHECKPOINT_LOCAL
    )

    completed_images = set(
        checkpoint_df["image_path"].astype(str)
    )

    print(
        "Previously completed:",
        len(completed_images)
    )

else:

    checkpoint_df = pd.DataFrame()
    completed_images = set()

# ------------------------------------------------------------
# Remaining images
# ------------------------------------------------------------

remaining_images = [
    p for p in val_images
    if str(p) not in completed_images
]

print(
    "Total validation images:",
    len(val_images)
)

print(
    "Already completed:",
    len(completed_images)
)

print(
    "Remaining:",
    len(remaining_images)
)

if len(remaining_images) == 0:

    print("\n========== MINING ALREADY COMPLETE ==========")

else:

    # --------------------------------------------------------
    # Mining results
    # --------------------------------------------------------

    results_list = []

    start_time = time.time()

    for idx, image_path in enumerate(
        remaining_images,
        start=1
    ):

        image_start = time.time()

        try:

            # ================================================
            # YOLO inference
            # ================================================

            predictions = model.predict(
                source=str(image_path),
                conf=CONF_THRESHOLD,
                imgsz=IMAGE_SIZE,
                device=0,
                verbose=False
            )

            result = predictions[0]

            # ================================================
            # Image dimensions
            # ================================================

            image = cv2.imread(
                str(image_path)
            )

            if image is None:
                raise RuntimeError(
                    "Could not read image"
                )

            height, width = image.shape[:2]

            # ================================================
            # Predictions
            # ================================================

            pred_boxes = (
                result.boxes.xyxy
                .cpu()
                .numpy()
            )

            pred_conf = (
                result.boxes.conf
                .cpu()
                .numpy()
            )

            # ================================================
            # Ground truth
            # ================================================

            sequence_name = image_path.parent.name

            label_path = (
                VAL_LABELS
                / sequence_name
                / f"{image_path.stem}.txt"
            )

            gt_yolo = load_yolo_labels(
                label_path
            )

            gt_boxes = yolo_to_xyxy(
                gt_yolo,
                width,
                height
            )

            # ================================================
            # Matching
            # ================================================

            tp, fp, fn, matched_ious = (
                greedy_match(
                    gt_boxes,
                    pred_boxes,
                    IOU_THRESHOLD
                )
            )

            # ================================================
            # Metrics
            # ================================================

            precision = (
                tp / (tp + fp)
                if (tp + fp) > 0
                else 0.0
            )

            recall = (
                tp / (tp + fn)
                if (tp + fn) > 0
                else 0.0
            )

            mean_iou = (
                float(np.mean(matched_ious))
                if matched_ious
                else 0.0
            )

            mean_confidence = (
                float(np.mean(pred_conf))
                if len(pred_conf) > 0
                else 0.0
            )

            difficulty = (
                fp + FN_WEIGHT * fn
            )

            elapsed = (
                time.time() - image_start
            )

            # ================================================
            # Save result
            # ================================================

            row = {
                "image_path": str(image_path),
                "sequence": sequence_name,
                "frame": image_path.stem,
                "gt": len(gt_boxes),
                "predictions": len(pred_boxes),
                "tp": tp,
                "fp": fp,
                "fn": fn,
                "precision": precision,
                "recall": recall,
                "mean_iou": mean_iou,
                "mean_confidence": mean_confidence,
                "difficulty": difficulty,
                "inference_time_sec": elapsed
            }

            results_list.append(row)

        except Exception as e:

            print(
                f"\nERROR processing {image_path}"
            )

            print(
                "Error:",
                repr(e)
            )

            # Record failure rather than losing progress
            results_list.append({
                "image_path": str(image_path),
                "sequence": image_path.parent.name,
                "frame": image_path.stem,
                "gt": -1,
                "predictions": -1,
                "tp": -1,
                "fp": -1,
                "fn": -1,
                "precision": -1,
                "recall": -1,
                "mean_iou": -1,
                "mean_confidence": -1,
                "difficulty": -1,
                "inference_time_sec": -1
            })

        # ====================================================
        # Periodic checkpoint
        # ====================================================

        if (
            len(results_list) >= SAVE_EVERY
            or idx == len(remaining_images)
        ):

            new_df = pd.DataFrame(
                results_list
            )

            if CHECKPOINT_LOCAL.exists():

                old_df = pd.read_csv(
                    CHECKPOINT_LOCAL
                )

                combined_df = pd.concat(
                    [old_df, new_df],
                    ignore_index=True
                )

            else:

                combined_df = new_df

            # Remove accidental duplicates
            combined_df = (
                combined_df
                .drop_duplicates(
                    subset=["image_path"],
                    keep="last"
                )
                .sort_values("image_path")
            )

            # Save locally
            combined_df.to_csv(
                CHECKPOINT_LOCAL,
                index=False
            )

            combined_df.to_csv(
                RESULTS_LOCAL,
                index=False
            )

            # ----------------------------------------------
            # Persistent Drive checkpoint
            # ----------------------------------------------

            shutil.copy2(
                CHECKPOINT_LOCAL,
                CHECKPOINT_DRIVE
            )

            shutil.copy2(
                RESULTS_LOCAL,
                RESULTS_DRIVE
            )

            # Clear temporary results
            results_list = []

            # ----------------------------------------------
            # Progress
            # ----------------------------------------------

            completed_now = len(
                combined_df
            )

            total_elapsed = (
                time.time() - start_time
            )

            avg_time = (
                total_elapsed / idx
            )

            remaining = (
                len(remaining_images) - idx
            )

            eta_minutes = (
                remaining * avg_time / 60
            )

            print(
                f"\n[{completed_now}/{len(val_images)}]"
                f" | Remaining: {len(val_images)-completed_now}"
                f" | Avg: {avg_time:.2f}s/image"
                f" | ETA: {eta_minutes:.1f} min"
            )

    print(
        "\n========== HARD MINING FINISHED =========="
    )

# ============================================================
# Final summary
# ============================================================

if CHECKPOINT_LOCAL.exists():

    mining_df = pd.read_csv(
        CHECKPOINT_LOCAL
    )

    print("\nTotal results:", len(mining_df))

    valid_df = mining_df[
        mining_df["gt"] >= 0
    ]

    print(
        "Valid results:",
        len(valid_df)
    )

    print(
        "Failed results:",
        len(mining_df) - len(valid_df)
    )

    print("\nAggregate:")
    print(
        "GT:",
        int(valid_df["gt"].sum())
    )

    print(
        "Predictions:",
        int(valid_df["predictions"].sum())
    )

    print(
        "TP:",
        int(valid_df["tp"].sum())
    )

    print(
        "FP:",
        int(valid_df["fp"].sum())
    )

    print(
        "FN:",
        int(valid_df["fn"].sum())
    )

    total_tp = valid_df["tp"].sum()
    total_fp = valid_df["fp"].sum()
    total_fn = valid_df["fn"].sum()

    overall_precision = (
        total_tp /
        (total_tp + total_fp)
        if (total_tp + total_fp) > 0
        else 0
    )

    overall_recall = (
        total_tp /
        (total_tp + total_fn)
        if (total_tp + total_fn) > 0
        else 0
    )

    print(
        "Overall precision:",
        round(overall_precision, 4)
    )

    print(
        "Overall recall:",
        round(overall_recall, 4)
    )

    print(
        "\nResults saved to:"
    )

    print(
        RESULTS_DRIVE
    )
```

```python
# ============================================================
# CONFIDENCE + NMS SWEEP ON HARD FRAMES
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

# ------------------------------------------------------------
# Load mining results
# ------------------------------------------------------------

results_path = Path(
    "/content/drive/MyDrive/AthletiQ/sportsmot/"
    "experiments/yolov26m_sportsmot/mining_results.csv"
)

df = pd.read_csv(results_path)

# ------------------------------------------------------------
# Hard sequences selected from visual analysis
# ------------------------------------------------------------

target_sequences = [
    "v_G-vNjfx1GGc_c601",
    "v_G-vNjfx1GGc_c600",
    "v_ITo3sCnpw_k_c007",
    "v_G-vNjfx1GGc_c008",
    "v_dw7LOz17Omg_c053",
]

hard_df = df[
    df["sequence"].isin(target_sequences)
].copy()

# Select top 30 hardest frames from these sequences
hard_df = hard_df.sort_values(
    "difficulty",
    ascending=False
).head(30)

print("Hard frames used:", len(hard_df))

display(
    hard_df[
        [
            "sequence",
            "frame",
            "gt",
            "predictions",
            "tp",
            "fp",
            "fn",
            "difficulty"
        ]
    ]
)

# ------------------------------------------------------------
# Load GT
# ------------------------------------------------------------

def get_gt_boxes(image_path, sequence, frame):
    image = cv2.imread(str(image_path))

    if image is None:
        return None

    h, w = image.shape[:2]

    label_path = (
        VAL_LABELS
        / sequence
        / f"{int(frame):06d}.txt"
    )

    gt_yolo = load_yolo_labels(label_path)

    return yolo_to_xyxy(
        gt_yolo,
        w,
        h
    )


# ------------------------------------------------------------
# IoU function
# ------------------------------------------------------------

def box_iou(box_a, box_b):

    x1 = max(box_a[0], box_b[0])
    y1 = max(box_a[1], box_b[1])
    x2 = min(box_a[2], box_b[2])
    y2 = min(box_a[3], box_b[3])

    inter_w = max(0, x2 - x1)
    inter_h = max(0, y2 - y1)

    inter = inter_w * inter_h

    area_a = max(
        0,
        box_a[2] - box_a[0]
    ) * max(
        0,
        box_a[3] - box_a[1]
    )

    area_b = max(
        0,
        box_b[2] - box_b[0]
    ) * max(
        0,
        box_b[3] - box_b[1]
    )

    union = area_a + area_b - inter

    if union <= 0:
        return 0.0

    return inter / union


# ------------------------------------------------------------
# Match predictions to GT
# ------------------------------------------------------------

def evaluate_predictions(
    gt_boxes,
    pred_boxes,
    iou_threshold=0.50
):

    if len(gt_boxes) == 0:

        return (
            0,
            len(pred_boxes),
            0
        )

    if len(pred_boxes) == 0:

        return (
            0,
            0,
            len(gt_boxes)
        )

    matches = []

    for pi, pred in enumerate(pred_boxes):

        for gi, gt in enumerate(gt_boxes):

            iou = box_iou(
                pred,
                gt
            )

            if iou >= iou_threshold:

                matches.append(
                    (
                        iou,
                        pi,
                        gi
                    )
                )

    # Highest IoU first
    matches.sort(
        reverse=True
    )

    matched_preds = set()
    matched_gts = set()

    tp = 0

    for iou, pi, gi in matches:

        if pi in matched_preds:
            continue

        if gi in matched_gts:
            continue

        matched_preds.add(pi)
        matched_gts.add(gi)

        tp += 1

    fp = len(pred_boxes) - tp
    fn = len(gt_boxes) - tp

    return tp, fp, fn


# ------------------------------------------------------------
# Parameter grid
# ------------------------------------------------------------

confidence_values = [
    0.20,
    0.25,
    0.30,
    0.35,
    0.40,
    0.50
]

nms_values = [
    0.40,
    0.45,
    0.50,
    0.55,
    0.60
]

# ------------------------------------------------------------
# Run sweep
# ------------------------------------------------------------

sweep_results = []

for conf in confidence_values:

    for nms_iou in nms_values:

        total_gt = 0
        total_pred = 0
        total_tp = 0
        total_fp = 0
        total_fn = 0

        for _, row in hard_df.iterrows():

            image_path = Path(
                row["image_path"]
            )

            gt_boxes = get_gt_boxes(
                image_path,
                row["sequence"],
                row["frame"]
            )

            if gt_boxes is None:
                continue

            prediction = model.predict(
                source=str(image_path),
                conf=conf,
                iou=nms_iou,
                imgsz=IMAGE_SIZE,
                device=0,
                verbose=False
            )[0]

            pred_boxes = (
                prediction.boxes.xyxy
                .cpu()
                .numpy()
            )

            tp, fp, fn = evaluate_predictions(
                gt_boxes,
                pred_boxes,
                iou_threshold=0.50
            )

            total_gt += len(gt_boxes)
            total_pred += len(pred_boxes)

            total_tp += tp
            total_fp += fp
            total_fn += fn

        precision = (
            total_tp /
            (total_tp + total_fp)
            if total_tp + total_fp > 0
            else 0
        )

        recall = (
            total_tp /
            (total_tp + total_fn)
            if total_tp + total_fn > 0
            else 0
        )

        f1 = (
            2 * precision * recall /
            (precision + recall)
            if precision + recall > 0
            else 0
        )

        sweep_results.append({
            "confidence": conf,
            "nms_iou": nms_iou,
            "gt": total_gt,
            "predictions": total_pred,
            "tp": total_tp,
            "fp": total_fp,
            "fn": total_fn,
            "precision": precision,
            "recall": recall,
            "f1": f1
        })

        print(
            f"conf={conf:.2f} | "
            f"nms={nms_iou:.2f} | "
            f"TP={total_tp} | "
            f"FP={total_fp} | "
            f"FN={total_fn} | "
            f"P={precision:.4f} | "
            f"R={recall:.4f} | "
            f"F1={f1:.4f}"
        )


# ------------------------------------------------------------
# Results table
# ------------------------------------------------------------

sweep_df = pd.DataFrame(
    sweep_results
)

sweep_df = sweep_df.sort_values(
    "f1",
    ascending=False
)

print("\n========== BEST CONFIGURATIONS ==========")

display(
    sweep_df.head(15)
)

# ------------------------------------------------------------
# Save results
# ------------------------------------------------------------

save_path = (
    Path(
        "/content/drive/MyDrive/AthletiQ/sportsmot/"
        "experiments/yolov26m_sportsmot/"
        "confidence_nms_sweep.csv"
    )
)

sweep_df.to_csv(
    save_path,
    index=False
)

print("\nSaved:")
print(save_path)
```

```python
# ============================================================
# FULL SPORTSMOT VALIDATION
# CONFIDENCE + NMS COMPARISON
# ============================================================

import cv2
import pandas as pd
import numpy as np
from pathlib import Path
from time import time

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_ROOT = Path(
    "/content/AthletiQ"
)

DATASET_ROOT = (
    PROJECT_ROOT
    / "datasets/football/processed/sportsmot_yolo"
)

VAL_IMAGES_ROOT = (
    DATASET_ROOT / "images" / "val"
)

VAL_LABELS_ROOT = (
    DATASET_ROOT / "labels" / "val"
)

OUTPUT_ROOT = Path(
    "/content/drive/MyDrive/AthletiQ/sportsmot/"
    "experiments/yolov26m_sportsmot"
)

OUTPUT_ROOT.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# Configurations
# ------------------------------------------------------------

CONFIGS = [
    {
        "name": "baseline",
        "confidence": 0.25,
        "nms_iou": 0.50
    },
    {
        "name": "candidate_A",
        "confidence": 0.30,
        "nms_iou": 0.40
    },
    {
        "name": "candidate_B",
        "confidence": 0.40,
        "nms_iou": 0.40
    },
    {
        "name": "candidate_C",
        "confidence": 0.50,
        "nms_iou": 0.45
    }
]

IMAGE_SIZE = 1280
BATCH_SIZE = 8
MATCH_IOU = 0.50

# ------------------------------------------------------------
# Collect validation images recursively
# ------------------------------------------------------------

image_paths = sorted(
    VAL_IMAGES_ROOT.rglob("*.jpg")
)

print(
    f"Validation images found: {len(image_paths):,}"
)

assert len(image_paths) == 9058, (
    f"Expected 9058 validation images, "
    f"found {len(image_paths)}"
)

# ------------------------------------------------------------
# IoU
# ------------------------------------------------------------

def box_iou(box_a, box_b):

    x1 = max(box_a[0], box_b[0])
    y1 = max(box_a[1], box_b[1])

    x2 = min(box_a[2], box_b[2])
    y2 = min(box_a[3], box_b[3])

    inter_w = max(
        0,
        x2 - x1
    )

    inter_h = max(
        0,
        y2 - y1
    )

    inter = (
        inter_w *
        inter_h
    )

    area_a = (
        max(0, box_a[2] - box_a[0]) *
        max(0, box_a[3] - box_a[1])
    )

    area_b = (
        max(0, box_b[2] - box_b[0]) *
        max(0, box_b[3] - box_b[1])
    )

    union = (
        area_a +
        area_b -
        inter
    )

    if union <= 0:
        return 0.0

    return inter / union


# ------------------------------------------------------------
# Greedy GT/prediction matching
# ------------------------------------------------------------

def match_boxes(
    gt_boxes,
    pred_boxes,
    threshold=0.50
):

    if len(gt_boxes) == 0:

        return (
            0,
            len(pred_boxes),
            0
        )

    if len(pred_boxes) == 0:

        return (
            0,
            0,
            len(gt_boxes)
        )

    candidates = []

    for pi, pred in enumerate(pred_boxes):

        for gi, gt in enumerate(gt_boxes):

            iou = box_iou(
                pred,
                gt
            )

            if iou >= threshold:

                candidates.append(
                    (
                        iou,
                        pi,
                        gi
                    )
                )

    candidates.sort(
        reverse=True
    )

    matched_predictions = set()
    matched_gt = set()

    tp = 0

    for iou, pi, gi in candidates:

        if pi in matched_predictions:
            continue

        if gi in matched_gt:
            continue

        matched_predictions.add(pi)
        matched_gt.add(gi)

        tp += 1

    fp = (
        len(pred_boxes) -
        tp
    )

    fn = (
        len(gt_boxes) -
        tp
    )

    return tp, fp, fn


# ------------------------------------------------------------
# Load YOLO GT labels
# ------------------------------------------------------------

def load_gt_for_image(image_path):

    image = cv2.imread(
        str(image_path)
    )

    if image is None:
        raise RuntimeError(
            f"Could not read {image_path}"
        )

    h, w = image.shape[:2]

    sequence = image_path.parent.name
    frame = image_path.stem

    label_path = (
        VAL_LABELS_ROOT
        / sequence
        / f"{frame}.txt"
    )

    if not label_path.exists():

        raise FileNotFoundError(
            f"Missing label: {label_path}"
        )

    gt_boxes = []

    with open(
        label_path,
        "r"
    ) as f:

        for line in f:

            parts = line.strip().split()

            if len(parts) < 5:
                continue

            # YOLO:
            # class cx cy width height

            cx = float(parts[1]) * w
            cy = float(parts[2]) * h
            bw = float(parts[3]) * w
            bh = float(parts[4]) * h

            x1 = cx - bw / 2
            y1 = cy - bh / 2
            x2 = cx + bw / 2
            y2 = cy + bh / 2

            gt_boxes.append(
                [
                    x1,
                    y1,
                    x2,
                    y2
                ]
            )

    return np.asarray(
        gt_boxes,
        dtype=np.float32
    )


# ------------------------------------------------------------
# Run one configuration
# ------------------------------------------------------------

def evaluate_configuration(
    config
):

    name = config["name"]
    confidence = config["confidence"]
    nms_iou = config["nms_iou"]

    print("\n" + "=" * 70)
    print(
        f"RUNNING: {name} | "
        f"conf={confidence:.2f} | "
        f"NMS IoU={nms_iou:.2f}"
    )
    print("=" * 70)

    total_gt = 0
    total_predictions = 0
    total_tp = 0
    total_fp = 0
    total_fn = 0

    failed = 0

    start_time = time()

    # Process batches
    for start in range(
        0,
        len(image_paths),
        BATCH_SIZE
    ):

        batch_paths = image_paths[
            start:start + BATCH_SIZE
        ]

        try:

            predictions = model.predict(
                source=[
                    str(p)
                    for p in batch_paths
                ],
                conf=confidence,
                iou=nms_iou,
                imgsz=IMAGE_SIZE,
                device=0,
                batch=BATCH_SIZE,
                verbose=False
            )

        except Exception as e:

            print(
                f"\nERROR in batch "
                f"{start}: {e}"
            )

            failed += len(
                batch_paths
            )

            continue

        for image_path, result in zip(
            batch_paths,
            predictions
        ):

            try:

                gt_boxes = (
                    load_gt_for_image(
                        image_path
                    )
                )

                pred_boxes = (
                    result.boxes.xyxy
                    .detach()
                    .cpu()
                    .numpy()
                )

                tp, fp, fn = (
                    match_boxes(
                        gt_boxes,
                        pred_boxes,
                        MATCH_IOU
                    )
                )

                total_gt += len(
                    gt_boxes
                )

                total_predictions += len(
                    pred_boxes
                )

                total_tp += tp
                total_fp += fp
                total_fn += fn

            except Exception as e:

                print(
                    f"\nERROR processing "
                    f"{image_path}: {e}"
                )

                failed += 1

        # Progress every 500 images
        processed = min(
            start + BATCH_SIZE,
            len(image_paths)
        )

        if (
            processed % 500 < BATCH_SIZE
            or processed == len(image_paths)
        ):

            elapsed = time() - start_time

            print(
                f"Progress: "
                f"{processed:,}/"
                f"{len(image_paths):,} | "
                f"elapsed={elapsed/60:.1f} min"
            )

    precision = (
        total_tp /
        (total_tp + total_fp)
        if total_tp + total_fp > 0
        else 0
    )

    recall = (
        total_tp /
        (total_tp + total_fn)
        if total_tp + total_fn > 0
        else 0
    )

    f1 = (
        2 * precision * recall /
        (precision + recall)
        if precision + recall > 0
        else 0
    )

    elapsed = time() - start_time

    result = {
        "name": name,
        "confidence": confidence,
        "nms_iou": nms_iou,
        "images": len(image_paths),
        "failed": failed,
        "gt": total_gt,
        "predictions": total_predictions,
        "tp": total_tp,
        "fp": total_fp,
        "fn": total_fn,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "time_minutes": elapsed / 60
    }

    print("\nRESULT")
    print("-" * 70)

    for key, value in result.items():
        print(
            f"{key:20s}: {value}"
        )

    return result


# ------------------------------------------------------------
# Run all configurations
# ------------------------------------------------------------

all_results = []

for config in CONFIGS:

    result = evaluate_configuration(
        config
    )

    all_results.append(
        result
    )

    # Save after EVERY configuration
    # so a runtime disconnect does not
    # destroy completed results.

    partial_df = pd.DataFrame(
        all_results
    )

    partial_path = (
        OUTPUT_ROOT
        / "full_validation_threshold_sweep.csv"
    )

    partial_df.to_csv(
        partial_path,
        index=False
    )

    print(
        f"\nSaved partial results to:\n"
        f"{partial_path}"
    )


# ------------------------------------------------------------
# Final comparison
# ------------------------------------------------------------

results_df = pd.DataFrame(
    all_results
)

results_df = results_df.sort_values(
    "f1",
    ascending=False
)

print("\n")
print("=" * 80)
print("FINAL FULL-VALIDATION COMPARISON")
print("=" * 80)

display(
    results_df[
        [
            "name",
            "confidence",
            "nms_iou",
            "gt",
            "predictions",
            "tp",
            "fp",
            "fn",
            "precision",
            "recall",
            "f1",
            "failed",
            "time_minutes"
        ]
    ]
)

# ------------------------------------------------------------
# Save final results
# ------------------------------------------------------------

final_path = (
    OUTPUT_ROOT
    / "full_validation_threshold_sweep.csv"
)

results_df.to_csv(
    final_path,
    index=False
)

print(
    "\nFinal results saved to:"
)

print(final_path)
```

```python
# ============================================================
# ATHLETIQ MODEL 1
# BUILD TARGETED HARD-EXAMPLE FINE-TUNING DATASET
# ============================================================

import pandas as pd
import numpy as np
import shutil
from pathlib import Path

# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

PROJECT_ROOT = Path("/content/AthletiQ")

SOURCE_IMAGES = (
    PROJECT_ROOT
    / "datasets/football/processed/sportsmot_yolo/images/val"
)

SOURCE_LABELS = (
    PROJECT_ROOT
    / "datasets/football/processed/sportsmot_yolo/labels/val"
)

MINING_RESULTS = Path(
    "/content/drive/MyDrive/AthletiQ/sportsmot/"
    "experiments/yolov26m_sportsmot/mining_results.csv"
)

OUTPUT_ROOT = (
    PROJECT_ROOT
    / "datasets/football/processed/"
    "sportsmot_hard_finetune"
)

OUTPUT_IMAGES = OUTPUT_ROOT / "images/train"
OUTPUT_LABELS = OUTPUT_ROOT / "labels/train"

OUTPUT_IMAGES.mkdir(
    parents=True,
    exist_ok=True
)

OUTPUT_LABELS.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# Load mining results
# ------------------------------------------------------------

df = pd.read_csv(
    MINING_RESULTS
)

print(
    "Total mining records:",
    len(df)
)

# ------------------------------------------------------------
# Target sequences
# ------------------------------------------------------------

target_sequences = [
    "v_G-vNjfx1GGc_c601",
    "v_G-vNjfx1GGc_c600",
    "v_ITo3sCnpw_k_c007",
    "v_dw7LOz17Omg_c053",
    "v_G-vNjfx1GGc_c008",
]

# IMPORTANT:
# c006 is deliberately excluded because of
# the previously discovered duplicate GT annotations.

df = df[
    df["sequence"].isin(
        target_sequences
    )
].copy()

print(
    "Records in target sequences:",
    len(df)
)

# ------------------------------------------------------------
# Create difficulty score specifically
# for fine-tuning selection
#
# FN is weighted more heavily because
# missed players are especially harmful
# for downstream tracking.
# ------------------------------------------------------------

df["hard_score"] = (
    df["fp"] +
    2.0 * df["fn"]
)

# ------------------------------------------------------------
# Select:
#
# 1. Highest hard_score
# 2. FN-heavy frames
# 3. FP-heavy frames
#
# Per sequence, keep a controlled number
# rather than letting one sequence dominate.
# ------------------------------------------------------------

selected = []

MAX_PER_SEQUENCE = 120

for sequence in target_sequences:

    seq_df = df[
        df["sequence"] == sequence
    ].copy()

    if len(seq_df) == 0:
        continue

    # Sort by hard score
    seq_df = seq_df.sort_values(
        "hard_score",
        ascending=False
    )

    # Take top difficult frames
    top_frames = seq_df.head(
        MAX_PER_SEQUENCE
    )

    selected.append(
        top_frames
    )

selected_df = pd.concat(
    selected,
    ignore_index=True
)

# Remove duplicate image paths
selected_df = selected_df.drop_duplicates(
    subset=["image_path"]
)

print(
    "\nSelected hard frames:",
    len(selected_df)
)

# ------------------------------------------------------------
# Print sequence distribution
# ------------------------------------------------------------

print("\nSequence distribution:")

print(
    selected_df["sequence"]
    .value_counts()
)

# ------------------------------------------------------------
# Print error statistics
# ------------------------------------------------------------

print("\nSelected dataset statistics:")

print(
    "GT:",
    int(selected_df["gt"].sum())
)

print(
    "Predictions:",
    int(selected_df["predictions"].sum())
)

print(
    "TP:",
    int(selected_df["tp"].sum())
)

print(
    "FP:",
    int(selected_df["fp"].sum())
)

print(
    "FN:",
    int(selected_df["fn"].sum())
)

# ------------------------------------------------------------
# Copy images and labels
# ------------------------------------------------------------

copied = 0
missing = 0

manifest = []

for _, row in selected_df.iterrows():

    sequence = row["sequence"]
    frame = str(row["frame"])

    # Original image path
    source_image = Path(
        row["image_path"]
    )

    # Label path
    source_label = (
        SOURCE_LABELS
        / sequence
        / f"{int(row['frame']):06d}.txt"
    )

    if not source_image.exists():

        missing += 1
        continue

    if not source_label.exists():

        missing += 1
        continue

    # Unique filename prevents collisions
    filename = (
        f"{sequence}__"
        f"{int(row['frame']):06d}.jpg"
    )

    label_filename = (
        f"{sequence}__"
        f"{int(row['frame']):06d}.txt"
    )

    destination_image = (
        OUTPUT_IMAGES / filename
    )

    destination_label = (
        OUTPUT_LABELS / label_filename
    )

    shutil.copy2(
        source_image,
        destination_image
    )

    shutil.copy2(
        source_label,
        destination_label
    )

    manifest.append({
        "sequence": sequence,
        "frame": int(row["frame"]),
        "source_image": str(source_image),
        "hard_score": row["hard_score"],
        "fp": int(row["fp"]),
        "fn": int(row["fn"]),
        "difficulty": row["difficulty"],
        "new_image": str(destination_image)
    })

    copied += 1

# ------------------------------------------------------------
# Save manifest
# ------------------------------------------------------------

manifest_df = pd.DataFrame(
    manifest
)

manifest_path = (
    OUTPUT_ROOT
    / "hard_finetune_manifest.csv"
)

manifest_df.to_csv(
    manifest_path,
    index=False
)

# ------------------------------------------------------------
# data.yaml
# ------------------------------------------------------------

yaml_path = (
    OUTPUT_ROOT
    / "data.yaml"
)

yaml_content = f"""
path: {OUTPUT_ROOT}
train: images/train

names:
  0: player
"""

yaml_path.write_text(
    yaml_content.strip()
)

# ------------------------------------------------------------
# Final report
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("HARD FINE-TUNING DATASET READY")
print("=" * 70)

print(
    f"Copied images : {copied}"
)

print(
    f"Missing files : {missing}"
)

print(
    f"Labels        : "
    f"{len(list(OUTPUT_LABELS.glob('*.txt')))}"
)

print(
    "\nDataset:"
)

print(
    OUTPUT_ROOT
)

print(
    "\nYAML:"
)

print(
    yaml_path
)

print(
    "\nManifest:"
)

print(
    manifest_path
)
```

```python
# ============================================================
# ATHLETIQ MODEL 1 — START FINE-TUNING
# YOLOv26m SportsMOT Football
# ============================================================

from pathlib import Path
from ultralytics import YOLO
import torch

# ------------------------------------------------------------
# 1. Paths
# ------------------------------------------------------------

PROJECT_ROOT = Path("/content/AthletiQ")

MODEL_PATH = (
    PROJECT_ROOT
    / "ai-engine/models/"
    "yolov26m-sportsmot-football.pt"
)

DATA_YAML = (
    PROJECT_ROOT
    / "datasets/football/processed/"
    "sportsmot_hard_finetune/data.yaml"
)

DRIVE_OUTPUT = Path(
    "/content/drive/MyDrive/AthletiQ/sportsmot/"
    "experiments/yolov26m_sportsmot/fine_tuning"
)

DRIVE_OUTPUT.mkdir(
    parents=True,
    exist_ok=True
)

# ------------------------------------------------------------
# 2. Verify GPU
# ------------------------------------------------------------

print("=" * 70)
print("ATHLETIQ MODEL 1 — FINE-TUNING")
print("=" * 70)

print("PyTorch:", torch.__version__)
print("CUDA available:", torch.cuda.is_available())

if not torch.cuda.is_available():
    raise RuntimeError(
        "CUDA GPU is not available. "
        "Switch Colab to a GPU runtime before training."
    )

print(
    "GPU:",
    torch.cuda.get_device_name(0)
)

print(
    "VRAM:",
    round(
        torch.cuda.get_device_properties(0).total_memory
        / 1024**3,
        2
    ),
    "GB"
)

# ------------------------------------------------------------
# 3. Verify model and dataset
# ------------------------------------------------------------

if not MODEL_PATH.exists():
    raise FileNotFoundError(
        f"Model not found:\n{MODEL_PATH}"
    )

if not DATA_YAML.exists():
    raise FileNotFoundError(
        f"Dataset YAML not found:\n{DATA_YAML}"
    )

print("\nModel:")
print(MODEL_PATH)

print("\nDataset YAML:")
print(DATA_YAML)

print("\nDataset YAML contents:")
print(DATA_YAML.read_text())

# ------------------------------------------------------------
# 4. Load existing SportsMOT football model
# ------------------------------------------------------------

model = YOLO(
    str(MODEL_PATH)
)

print("\nModel loaded successfully.")

print("Task:", model.task)
print("Classes:", model.names)

# ------------------------------------------------------------
# 5. Fine-tuning configuration
# ------------------------------------------------------------

EPOCHS = 20
IMAGE_SIZE = 1280
BATCH_SIZE = 4

# ------------------------------------------------------------
# 6. Start fine-tuning
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("STARTING TRAINING")
print("=" * 70)

results = model.train(

    # Dataset
    data=str(DATA_YAML),

    # Training
    epochs=EPOCHS,
    imgsz=IMAGE_SIZE,
    batch=BATCH_SIZE,

    # GPU
    device=0,

    # Data loading
    workers=2,
    cache=False,

    # Mixed precision for T4
    amp=True,

    # Early stopping
    patience=5,

    # Save checkpoints
    save=True,
    save_period=1,

    # Output
    project=str(DRIVE_OUTPUT),
    name="yolov26m_hard_finetune",

    # Start from our existing trained model
    pretrained=str(MODEL_PATH),

    # Let Ultralytics select optimizer
    optimizer="auto",

    # Reproducibility
    seed=42,
    deterministic=True,

    # Validate during training
    val=True,

    # Logging
    verbose=True
)

# ------------------------------------------------------------
# 7. Training finished
# ------------------------------------------------------------

TRAIN_DIR = (
    DRIVE_OUTPUT /
    "yolov26m_hard_finetune"
)

BEST_MODEL = (
    TRAIN_DIR /
    "weights" /
    "best.pt"
)

LAST_MODEL = (
    TRAIN_DIR /
    "weights" /
    "last.pt"
)

print("\n" + "=" * 70)
print("TRAINING FINISHED")
print("=" * 70)

print("\nTraining directory:")
print(TRAIN_DIR)

print("\nBest model:")
print(BEST_MODEL)

print("\nLast model:")
print(LAST_MODEL)

print("\nBest model exists:", BEST_MODEL.exists())
print("Last model exists:", LAST_MODEL.exists())
```
